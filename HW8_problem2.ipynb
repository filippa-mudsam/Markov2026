```python
import numpy as np
from math import factorial
```

```python
rng = np.random.default_rng()
```

```python
#variables
nsim = int(2*10**5)
t=5
lams=3
lamk = 2
q = 0.4
qlams = q * lams
qlamk = q* lamk
```

```python
# simulate w/ prob q
Ns = rng.poisson(lams*t, nsim)
Nk = rng.poisson(lamk*t, nsim)
Ds = rng.binomial(Ns,q)
Dk = rng.binomial(Nk, q)

#simulate directly
Ds_direct = rng.poisson(qlams*t, nsim)
Dk_direct = rng.poisson(qlamk*t, nsim)
```

```python
print('method 1: star mean', np.mean(Ds), 'var', np.var(Ds),
      '  sky mean', np.mean(Dk), 'var', np.var(Dk))
print('method 2: star mean', np.mean(Ds_direct), 'var', np.var(Ds_direct),
      '  sky mean', np.mean(Dk_direct), 'var', np.var(Dk_direct))
 
# correlation between detected star and sky counts
print('corr(star, sky): method 1', np.corrcoef(Ds, Dk)[0, 1],
      '  method 2', np.corrcoef(Ds_direct, Dk_direct)[0, 1])
```

```python
#comparing detections w/ poisson(10)
D1 = Ds + Dk
D2 = Ds_direct + Dk_direct
lam = (qlams + qlamk)*t
k_r = np.arange(0,31)
f_poisson = np.zeros(len(k_r))

for j, k in enumerate(k_r):
    f_poisson[j] = np.exp(-lam)*float(lam)**k / factorial(k) #poisson pdf

emp1 = np.bincount(D1, minlength=len(k_r))[:len(k_r)]/nsim
emp2 = np.bincount(D2, minlength=len(k_r))[:len(k_r)]/nsim
```

```python

print('total: method 1 mean', np.mean(D1), 'var', np.var(D1),
      '  method 2 mean', np.mean(D2), 'var', np.var(D2))
print('max |pmf - Poisson(10) pmf|: method 1', np.max(np.abs(emp1 - f_poisson)),
      '  method 2', np.max(np.abs(emp2 - f_poisson)))
```

```python
ten = (D1 == 10)
print('P(6 star | 10 detections):', np.mean(Ds[ten] == 6))
 
TS = -np.log(rng.random(nsim))/qlams
TK = -np.log(rng.random(nsim))/qlamk
V = np.minimum(TS, TK)              
I = (TS < TK)                      
M = np.maximum(TS, TK)              
 
print('P(star first):', np.mean(I))
print('P(star first and within 1 s):', np.mean(I & (V <= 1)))
print('E[M]:', np.mean(M))
print('corr(time of first detection, type):', np.corrcoef(V, I)[0, 1])
```
