For m = 3, . . . , 12 find the smallest M with pM > 0 (track only the zero pattern, B ← (BP > 0)) and the second largest eigenvalue modulus |λ2|. Compare M with (m − 1)2 + 1, Wielandt’s bound on the exponent of any primitive m × m matrix. What does |λ2| say about qn → π as m grows?

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
ms = np.arange(3,13)
Ms = np.zeros(len(ms))
lambda_2 = np.zeros(len(ms))
```

```python
for k in range(len(ms)):
    m = ms[k]

    P=np.zeros((m,m))
    P= P + np.diag(np.ones(m-1), 1)
    P[m-1,0] = 1/2
    P[m-1, 1] = 1/2

    B = P > 0
    M = 1

    while not np.all(B):
        B = (B @ P) > 0 
        M = M + 1
    Ms[k] = M

    lambda_2[k] = np.sort(np.abs(np.linalg.eigvals(P)))[-2]

    print('m = %d   M = %d   bound = %d   |lambda 2| = %.5f' % (m, M, (m-1)**2 + 1, lambda_2[k]))
```

```python
lambda_2
```

```python
plt.figure(1, figsize=(8, 6))
plt.plot(ms, (ms-1)**2 + 1, 'r--', linewidth=3)
plt.plot(ms, Ms, 'black', linewidth=2)
plt.xlabel('m', fontsize=24)
plt.ylabel('M', fontsize=24)
plt.show()
```

```python
plt.figure(2, figsize=(8, 6))
plt.plot(ms, lambda_2, 'b', linewidth=3)
plt.xlabel('m', fontsize=24)
plt.ylabel('|lambda_2|', fontsize=24)
plt.show()
```

For the first plot, every m from 3 to 12 the M found here equals (m-1)^2 + 1 which is shown by the two curves perfectly overlapping eachother. For the second plot we can see that as m increases |lamba2| converges to 1. This means that as lambda2 gets closer to 1 the slower the chain will converge to the standard distribution.
