Let X, Y, and Z be independent, Uniform[0, 1] random variables.
(a) Analytically compute the probability that
X^2+Y^2<Z and Z^2>XY. 

Describe the region of the unit cube satisfying both inequalities and set up the integrals. (Hint: polar coordinates in x,y. The two lower bounds on z exchange dominance at some r∗(θ).) 

(b) Estimate the same probability by Monte Carlo and compare. Plot the estimate against sample size N with a logarithmic horizontal axis, and draw the analytic value as a horizontal line.


```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
Nsim = [10, 100, 1000, 10000]  #num of samples for empirical mean
analytic_answer = 23*np.pi/192
randn = np.random.default_rng()
probmc_sim = np.zeros(len(Nsim))
```

```python
#simulate 3 independant uniform RV's
for j, N in enumerate(Nsim):
    X = randn.uniform(0,1,N)
    Y = randn.uniform(0,1,N)
    Z = randn.uniform(0,1,N)

    #Regions or Conditions defined in problem
    region_1 = (X**2 + Y**2) < Z
    region_2 = (Z**2) > (X * Y)


    probmc_sim[j] = np.mean(region_1 & region_2)
```

```python
plt.plot([min(Nsim), max(Nsim)], [analytic_answer, analytic_answer], 'b', linewidth=5, label='analytic value')
plt.plot(Nsim, probmc_sim, 'r.', markersize=25, label='Monte Carlo estimate')
plt.xscale('log')
```
